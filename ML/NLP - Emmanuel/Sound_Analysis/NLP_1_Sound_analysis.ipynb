## Harmonics

We can use tone method to create wave sound given a frequency in hertz, also we can merge them to reproduce them together. in this example we are creating a fundamental sound and its harmonies.

```python
y = librosa.tone(400, duration=10)
y_harmonic = librosa.tone(800, duration=10)
y_harmonic2 = librosa.tone(1200, duration=10)
Audio(y+y_harmonic+y_harmonic2,rate=22100)
```

```python
plt.figure(figsize=(12, 4))
librosa.display.waveshow(y[:200])
librosa.display.waveshow(y_harmonic[:200])
librosa.display.waveshow(y_harmonic2[:200])
plt.show()
```
