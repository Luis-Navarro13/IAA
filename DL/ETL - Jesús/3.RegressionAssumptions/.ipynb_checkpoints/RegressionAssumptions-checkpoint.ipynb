```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_regression
```

```python
X, y = make_regression(n_samples = 10000, n_features = 1, noise = 20, random_state = 42)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
residuals =  y_test - y_pred
```

```python
plt.scatter(X_test, residuals)
plt.axhline(y=0, color='r',linestyle='--')
plt.xlabel('X_test')
plt.ylabel('Residuals')
```

```python
plt.figure(figsize=(5,5))
sns.scatterplot(x=X_test.flatten(),y=y_test, label="Current", color='blue')
sns.scatterplot(x=X_test.flatten(),y=y_pred, label="Predicted", color='red')
plt.xlabel('X_test: Indepent variable')
plt.ylabel('y_test: Dependent variable')
plt.title('Regression Line')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(5,5))
plt.scatter(x=y_pred,y=np.abs(residuals))
plt.axhline(y=0, color='r',linestyle='--')
plt.title('Homoscedasticity Plot')
plt.xlabel('Predicted values')
plt.ylabel('Absolute Residuals')
```

```python
np.random.seed(42)
X_example = np.random.rand(200, 1) * 100
y_example = 2 * X_example.squeeze() + np.random.normal(0,X_example.squeeze(), 200)

X_train, X_test, y_train, y_test = train_test_split(X_example, y_example, test_size=0.3, random_state=42)

model_h = LinearRegression()
model_h.fit(X_train, y_train)
y_pred_h = model_h.predict(X_test)
```

```python
residuals_h = y_test - y_pred_h
plt.scatter(y_pred_h, residuals_h)
plt.axhline(y=0, color='r', linestyle='--')
```

```python
plt.figure(figsize=(5,5))
plt.scatter(x=y_pred_h,y=np.abs(residuals_h))
plt.axhline(y=0, color='r',linestyle='--')
plt.title('Homoscedasticity Plot')
plt.xlabel('Predicted values')
plt.ylabel('Absolute Residuals')
plt.show()
```

```python
plt.figure(figsize=(5,5))
sns.histplot(residuals, kde=True)
plt.title('Residuals Plot')
plt.xlabel('Residuals')
plt.ylabel('Histograms of Residuals')
plt.show()
```

```python
plt.figure(figsize=(5,5))
sm.qqplot(residuals, line='s')
plt.xlabel('Theoretical Quanties')
plt.ylabel('Sample Quanties')
plt.title('QQ Plot of Residuals')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Load the dataset
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv"
wine_data = pd.read_csv(url, delimiter=";")
wine_data.head()
```

```python
X = np.array(wine_data['pH']).reshape(-1, 1)
y = np.array(wine_data['alcohol']).reshape(-1, 1)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
residuals =  y_test - y_pred
```

```python
plt.scatter(X_test, residuals)
plt.axhline(y=0, color='r',linestyle='--')
plt.xlabel('X_test')
plt.ylabel('Residuals')
```

```python
plt.scatter(X_test, residuals)
plt.axhline(y=0, color='r',linestyle='--')
plt.xlabel('X_test')
plt.ylabel('Residuals')
```
