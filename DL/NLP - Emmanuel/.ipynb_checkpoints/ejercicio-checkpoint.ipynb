```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time
from tqdm import tqdm  # pip install tqdm
```

```python
def get_data():
    n = 10 #NUMBER OF DIALOGUES
    request = "https://officeapi.akashrajpurohit.com/quote/"
    list_dialogues = [] 
    for i in tqdm(range(1, n + 1), desc="Obteniendo citas", unit="cita"):
        response = requests.get(request+str(i))
        response.encoding = 'utf-8'  # Forzar UTF-8
        soup = BeautifulSoup(response.content, 'html.parser')
        list_dialogues.append(soup)
        time.sleep(1.5)
    return list_dialogues
dialogues = get_data()
```
