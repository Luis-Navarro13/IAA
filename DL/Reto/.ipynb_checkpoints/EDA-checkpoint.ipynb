```python
# ======================================================
# EDA COMPLETO DE CARRERAS (Google Colab) - VERSIÓN CORREGIDA Y AMPLIADA
# ======================================================
# Qué hace:
# - Sube tu CSV y lo analiza automáticamente (columnas en inglés/español)
# - Limpia números con formatos tipo $47.375, 191,62%, 337.269, etc.
# - Graficas clave:
#     * Carreras más/menos populares (por total_students y por conteos)
#     * Carreras más baratas/caras (públicas, privadas y costo combinado)
#     * Distribuciones y boxplots (salarios, costos)
#     * Brecha salarial mujeres vs hombres (por carrera)
#     * Relación costo vs salario
#     * Tasa de empleo vs desempleo, informalidad, etc.
#     * Heatmap de correlaciones
#     * Nube de palabras de carreras
# - Exporta tablas (CSV) con rankings e insights
# ======================================================

# ---- Instalación de librerías ----
!pip install -q pandas numpy matplotlib seaborn plotly wordcloud unidecode

# ---- Imports ----
import io, re, os, numpy as np, pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
from wordcloud import WordCloud
from unidecode import unidecode
from google.colab import files

plt.rcParams['figure.dpi'] = 110
sns.set(style="whitegrid", context="notebook")

# ======================================================
# 1) SUBIR Y LEER ARCHIVO
# ======================================================
print("📤 Sube tu archivo CSV (ej: data_avance.csv)")
uploaded = files.upload()
assert len(uploaded) >= 1, "No se subió ningún archivo."
csv_name = list(uploaded.keys())[0]
print(f"✅ Archivo recibido: {csv_name}")

df = pd.read_csv(io.BytesIO(uploaded[csv_name]))
print("Dimensiones:", df.shape)
display(df.head())

# ======================================================
# 2) NORMALIZACIÓN DE COLUMNAS Y STRINGS
# ======================================================
orig_cols = df.columns.tolist()
df.columns = [unidecode(c.strip().lower().replace(' ', '_')) for c in df.columns]

for c in df.select_dtypes(include=['object']).columns:
    df[c] = df[c].astype(str).str.strip()

# ======================================================
# 3) DETECTOR DE COLUMNAS (sinónimos)
# ======================================================
def pick_col(candidates, df=df):
    cols = df.columns.tolist()
    for pat in candidates:
        if pat in cols:
            return pat
        for c in cols:
            if re.search(pat, c):
                return c
    return None

# Nombre de la carrera
col_carrera = pick_col([r'^carrera$', r'^name$', r'programa', r'licenciatura', r'ingenieria', r'nombre.*carrera'])
# Área/cluster
col_area = pick_col([r'^career$', r'^area$', r'campo', r'disciplina', r'categoria', r'facultad'])
# Institución (si existiera)
col_institucion = pick_col([r'institucion', r'universidad', r'escuela', r'campus'])
# Costos público/privado
col_costo_pub = pick_col([r'public_cost', r'costo_publico', r'colegiatura_publica'])
col_costo_priv = pick_col([r'private_cost', r'costo_privado', r'colegiatura_privada'])
# Popularidad (estudiantes)
col_total_students = pick_col([r'^total_students$', r'estudiantes_totales', r'matricula', r'alumnos'])
# Egresados recientes
col_new_grads = pick_col([r'new_graduates_number', r'nuevos_egresados', r'egresados'])
# Empleo/Desempleo/Informalidad
col_emp_rate = pick_col([r'^emp_rate$', r'tasa_empleo', r'employment_rate'])
col_unemp_rate = pick_col([r'^unemployment_rate$', r'tasa_desempleo'])
col_informality = pick_col([r'^informality_rate$', r'tasa_informalidad'])
col_quality_employment = pick_col([r'^quality_employment_probability$', r'probabilidad_empleo_calidad'])
# Salarios
col_avg_salary = pick_col([r'^average_salary$', r'salario_promedio', r'ingreso_promedio'])
col_women_salary = pick_col([r'^women_salary$', r'salario_mujeres'])
col_men_salary = pick_col([r'^men_salary$', r'salario_hombres'])
col_under30_salary = pick_col([r'^under_30_salary$', r'salario_menores_30'])
col_over30_salary = pick_col([r'^over_30_salary$', r'salario_mayores_30'])
col_formal_salary = pick_col([r'^formal_salary$', r'salario_formal'])
col_informal_salary = pick_col([r'^informal_salary$', r'salario_informal'])
col_postgrad_pct = pick_col([r'^postgrad_percentage$', r'porcentaje_posgrado'])
col_postgrad_salary = pick_col([r'^postgrad_salary$', r'salario_posgrado'])
col_salary_increase = pick_col([r'^salary_increase$', r'aumento_salarial'])
col_rank = pick_col([r'^career_rank$', r'ranking_carrera'])

print("\n🔎 Columnas detectadas:")
for k,v in {
    "carrera":col_carrera, "area":col_area, "institucion":col_institucion,
    "public_cost":col_costo_pub, "private_cost":col_costo_priv,
    "total_students":col_total_students, "new_graduates_number":col_new_grads,
    "emp_rate":col_emp_rate, "unemployment_rate":col_unemp_rate, "informality_rate":col_informality,
    "quality_employment_probability":col_quality_employment,
    "average_salary":col_avg_salary, "women_salary":col_women_salary, "men_salary":col_men_salary,
    "under_30_salary":col_under30_salary, "over_30_salary":col_over30_salary,
    "formal_salary":col_formal_salary, "informal_salary":col_informal_salary,
    "postgrad_percentage":col_postgrad_pct, "postgrad_salary":col_postgrad_salary,
    "salary_increase":col_salary_increase, "career_rank":col_rank
}.items():
    print(f" - {k:30s}: {v}")

# ======================================================
# 4) PARSEO NUMÉRICO ROBUSTO ($, %, miles con punto, decimales con coma)
# ======================================================
def to_num(series):
    s = series.astype(str)
    s = s.str.replace(r'[\$\s]', '', regex=True)
    s = s.str.replace('%', '', regex=False)
    # Caso con '.' y ',' -> miles con '.' y decimales con ','
    has_both = s.str.contains(r'\.') & s.str.contains(',')
    s = np.where(has_both, s.str.replace('.', '', regex=False).str.replace(',', '.', regex=False), s)
    s = pd.Series(s)
    # Sólo '.' como separador de miles (p.ej. 337.269)
    s = s.str.replace(r'(?<=\d)\.(?=\d{3}(\D|$))', '', regex=True)
    # Decimales con coma
    s = s.str.replace(',', '.', regex=False)
    return pd.to_numeric(s, errors='coerce')

# Lista de columnas numéricas esperadas
num_candidates = [
    col_costo_pub, col_costo_priv, col_total_students, col_new_grads,
    col_emp_rate, col_unemp_rate, col_informality, col_quality_employment,
    col_avg_salary, col_women_salary, col_men_salary,
    col_under30_salary, col_over30_salary,
    col_formal_salary, col_informal_salary,
    col_postgrad_pct, col_postgrad_salary, col_salary_increase
]
for c in [c for c in num_candidates if c]:
    df[c] = to_num(df[c])

# Costo combinado (promedio de público/privado cuando existan)
df['cost_combined'] = np.nan
if col_costo_pub and col_costo_priv:
    df['cost_combined'] = df[[col_costo_pub, col_costo_priv]].mean(axis=1, skipna=True)
elif col_costo_pub:
    df['cost_combined'] = df[col_costo_pub]
elif col_costo_priv:
    df['cost_combined'] = df[col_costo_priv]

# ======================================================
# 5) RESUMEN GLOBAL Y NULOS
# ======================================================
print("\n📋 RESUMEN GLOBAL")
print("Filas, columnas:", df.shape)
print("\nValores nulos por columna:\n", df.isna().sum())
display(df.describe(include='all').transpose())

# ======================================================
# 6) UTILIDADES DE PLOT
# ======================================================
def bar_top(series, n=15, title="", xlabel="", ylabel=""):
    vc = series.value_counts().head(n)
    plt.figure(figsize=(10,6))
    sns.barplot(x=vc.values, y=vc.index, palette="viridis")
    plt.title(title); plt.xlabel(xlabel); plt.ylabel(ylabel); plt.tight_layout(); plt.show()
    return vc

def bar_bottom(series, n=15, title="", xlabel="", ylabel=""):
    vc = series.value_counts().sort_values(ascending=True).head(n)
    plt.figure(figsize=(10,6))
    sns.barplot(x=vc.values, y=vc.index, palette="magma")
    plt.title(title); plt.xlabel(xlabel); plt.ylabel(ylabel); plt.tight_layout(); plt.show()
    return vc

def iqr_bounds(s):
    q1, q3 = s.quantile(0.25), s.quantile(0.75)
    iqr = q3 - q1
    return (q1 - 1.5*iqr, q3 + 1.5*iqr)

# ======================================================
# 7) POPULARIDAD / TENDENCIA
# ======================================================
if col_carrera:
    # Por conteos (número de registros por carrera)
    bar_top(df[col_carrera], n=20, title="Carreras más frecuentes (conteos)", xlabel="Registros", ylabel="Carrera")
    bar_bottom(df[col_carrera], n=20, title="Carreras menos frecuentes (conteos)", xlabel="Registros", ylabel="Carrera")

# Por total de estudiantes (mejor indicador de popularidad si existe)
if col_carrera and col_total_students:
    pop_by_students = df.groupby(col_carrera)[col_total_students].mean().sort_values(ascending=False)
    plt.figure(figsize=(10,6))
    sns.barplot(x=pop_by_students.head(20).values, y=pop_by_students.head(20).index, palette="cubehelix")
    plt.title("Carreras con más estudiantes (promedio total_students)"); plt.xlabel("Estudiantes (promedio)"); plt.ylabel("Carrera")
    plt.tight_layout(); plt.show()

    plt.figure(figsize=(10,6))
    sns.barplot(x=pop_by_students.tail(20).values, y=pop_by_students.tail(20).index, palette="flare")
    plt.title("Carreras con menos estudiantes (promedio total_students)"); plt.xlabel("Estudiantes (promedio)"); plt.ylabel("Carrera")
    plt.tight_layout(); plt.show()

# ======================================================
# 8) COSTOS: BARATAS/CARAS + DISTRIBUCIÓN + OUTLIERS + BOXES
# ======================================================
cost_col = 'cost_combined' if 'cost_combined' in df.columns else None

if cost_col and col_carrera:
    mean_cost = df.groupby(col_carrera)[cost_col].mean().dropna()
    top_costly = mean_cost.sort_values(ascending=False).head(20)
    top_cheap = mean_cost.sort_values(ascending=True).head(20)

    plt.figure(figsize=(10,6))
    sns.barplot(x=top_costly.values, y=top_costly.index, palette="Reds_r")
    plt.title("Carreras más costosas (promedio)"); plt.xlabel("Costo promedio"); plt.ylabel("Carrera"); plt.tight_layout(); plt.show()

    plt.figure(figsize=(10,6))
    sns.barplot(x=top_cheap.values, y=top_cheap.index, palette="Greens")
    plt.title("Carreras más baratas (promedio)"); plt.xlabel("Costo promedio"); plt.ylabel("Carrera"); plt.tight_layout(); plt.show()

    # Distribución de costos
    plt.figure(figsize=(9,5))
    sns.histplot(df[cost_col].dropna(), bins=40, kde=True)
    plt.title("Distribución de costos (combinado)"); plt.xlabel("Costo"); plt.tight_layout(); plt.show()

    # Outliers
    lo, hi = iqr_bounds(df[cost_col].dropna())
    out = df[(df[cost_col] < lo) | (df[cost_col] > hi)]
    print(f"\n🔔 Posibles outliers de costo (IQR): {len(out)}")
    display(out[[x for x in [col_carrera, cost_col, col_area] if x]].head(10))

    # Interactivo
    fig = px.histogram(df, x=cost_col, nbins=40, title="Distribución interactiva del costo (combinado)")
    fig.show()

# Boxplots por área si existe
if cost_col and col_area:
    plt.figure(figsize=(12,6))
    top_areas = df[col_area].value_counts().head(12).index
    sns.boxplot(data=df[df[col_area].isin(top_areas)], x=col_area, y=cost_col)
    plt.title(f"Costo por {col_area} (Top 12 áreas)")
    plt.xticks(rotation=45, ha='right'); plt.tight_layout(); plt.show()

# ======================================================
# 9) SALARIOS Y BRECHAS
# ======================================================
# Distribución de salarios clave
for sal_col in [col_avg_salary, col_women_salary, col_men_salary, col_formal_salary, col_informal_salary, col_postgrad_salary]:
    if sal_col:
        plt.figure(figsize=(9,5))
        sns.histplot(df[sal_col].dropna(), bins=40, kde=True)
        ttl = sal_col.replace('_',' ').title()
        plt.title(f"Distribución de {ttl}"); plt.xlabel("Monto"); plt.tight_layout(); plt.show()

# Brecha salarial mujeres-hombres por carrera
if col_carrera and col_women_salary and col_men_salary:
    gap = df.groupby(col_carrera)[[col_women_salary, col_men_salary]].mean().dropna()
    gap['gender_gap'] = gap[col_men_salary] - gap[col_women_salary]
    gap_sorted = gap.sort_values('gender_gap', ascending=False)
    plt.figure(figsize=(10,6))
    sns.barplot(x=gap_sorted['gender_gap'].head(20).values, y=gap_sorted.head(20).index, palette="coolwarm")
    plt.title("Brecha salarial (Hombres - Mujeres) por carrera (Top 20)"); plt.xlabel("Diferencia"); plt.ylabel("Carrera")
    plt.tight_layout(); plt.show()

# Relación costo vs salario promedio
if cost_col and col_avg_salary and col_carrera:
    agg = df.groupby(col_carrera)[[cost_col, col_avg_salary]].mean().dropna()
    plt.figure(figsize=(8,6))
    sns.scatterplot(data=agg, x=cost_col, y=col_avg_salary)
    plt.title("Costo vs Salario promedio (por carrera)"); plt.xlabel("Costo promedio"); plt.ylabel("Salario promedio")
    plt.tight_layout(); plt.show()

# Salarios <30 vs >30
if col_under30_salary and col_over30_salary:
    plt.figure(figsize=(8,6))
    sns.kdeplot(df[col_under30_salary].dropna(), label='<30', fill=True, alpha=0.4)
    sns.kdeplot(df[col_over30_salary].dropna(), label='≥30', fill=True, alpha=0.4)
    plt.title("Distribución de salarios por edad"); plt.xlabel("Salario"); plt.legend(); plt.tight_layout(); plt.show()

# ======================================================
# 10) EMPLEO / INFORMALIDAD / CALIDAD
# ======================================================
for rate_col, ttl in [
    (col_emp_rate, "Tasa de empleo"),
    (col_unemp_rate, "Tasa de desempleo"),
    (col_informality, "Tasa de informalidad"),
    (col_quality_employment, "Probabilidad de empleo de calidad")
]:
    if rate_col and col_carrera:
        rate_mean = df.groupby(col_carrera)[rate_col].mean().sort_values(ascending=False)
        plt.figure(figsize=(10,6))
        sns.barplot(x=rate_mean.head(15).values, y=rate_mean.head(15).index, palette="Blues")
        plt.title(f"Top 15 carreras por {ttl}"); plt.xlabel(ttl); plt.ylabel("Carrera"); plt.tight_layout(); plt.show()

# ======================================================
# 11) HEATMAP DE CORRELACIONES (numéricas)
# ======================================================
num_df = df.select_dtypes(include=[np.number]).copy()
if not num_df.empty and num_df.shape[1] >= 2:
    plt.figure(figsize=(10,7))
    corr = num_df.corr()
    sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm")
    plt.title("Mapa de calor de correlaciones numéricas"); plt.tight_layout(); plt.show()

# ======================================================
# 12) NUBE DE PALABRAS (CARRERAS)
# ======================================================
if col_carrera:
    text = " ".join(df[col_carrera].dropna().astype(str).tolist())
    text = unidecode(text.lower())
    wc = WordCloud(width=1000, height=400, background_color="white").generate(text)
    plt.figure(figsize=(12,5)); plt.imshow(wc); plt.axis("off"); plt.title("Nube de palabras - Carreras"); plt.show()

# ======================================================
# 13) EXPORTS DE TABLAS CLAVE
# ======================================================
exports = {}

# Popularidad por estudiantes
if col_carrera and col_total_students:
    pop_tbl = df.groupby(col_carrera)[col_total_students].mean().reset_index().sort_values(col_total_students, ascending=False)
    exports["pop_carreras_por_estudiantes.csv"] = pop_tbl

# Conteos por carrera (rareza)
if col_carrera:
    rare_tbl = df[col_carrera].value_counts().sort_values(ascending=True).reset_index()
    rare_tbl.columns = [col_carrera, 'conteo']
    exports["carreras_mas_raras.csv"] = rare_tbl

# Costos por carrera
if col_carrera and cost_col:
    cost_tbl = df.groupby(col_carrera)[cost_col].agg(['count','mean','median','min','max']).reset_index().sort_values('mean', ascending=False)
    exports["costos_por_carrera.csv"] = cost_tbl

# Salarios por carrera
if col_carrera and col_avg_salary:
    sal_tbl = df.groupby(col_carrera)[col_avg_salary].agg(['count','mean','median','min','max']).reset_index().sort_values('mean', ascending=False)
    exports["salarios_por_carrera.csv"] = sal_tbl

# Brecha salarial por carrera
if col_carrera and col_women_salary and col_men_salary:
    gap_tbl = df.groupby(col_carrera)[[col_women_salary, col_men_salary]].mean().dropna()
    gap_tbl['gender_gap'] = gap_tbl[col_men_salary] - gap_tbl[col_women_salary]
    gap_tbl = gap_tbl.reset_index().sort_values('gender_gap', ascending=False)
    exports["brecha_salarial_por_carrera.csv"] = gap_tbl

# Empleo/Desempleo/Informalidad por carrera
for rcol, name in [
    (col_emp_rate, "tasa_empleo_por_carrera.csv"),
    (col_unemp_rate, "tasa_desempleo_por_carrera.csv"),
    (col_informality, "tasa_informalidad_por_carrera.csv"),
    (col_quality_employment, "prob_empleo_calidad_por_carrera.csv"),
]:
    if rcol and col_carrera:
        t = df.groupby(col_carrera)[rcol].mean().reset_index().sort_values(rcol, ascending=False)
        exports[name] = t

# Guardar
for fname, table in exports.items():
    table.to_csv(fname, index=False)
print("\n💾 CSV generados:", list(exports.keys()))

# ======================================================
# 14) INSIGHTS RÁPIDOS
# ======================================================
def show_top(series_or_df, n=5):
    try:
        print(series_or_df.head(n).to_string())
    except:
        print(series_or_df.head(n))

print("\n📊 INSIGHTS")
print(f"- Registros: {len(df)} | Columnas: {df.shape[1]}")
if col_carrera:
    print(f"- Carreras distintas: {df[col_carrera].nunique()}")
if cost_col:
    print(f"- Costo combinado promedio: {df[cost_col].mean():,.2f} | mediana: {df[cost_col].median():,.2f}")
if col_avg_salary:
    print(f"- Salario promedio general: {df[col_avg_salary].mean():,.2f}")
if col_carrera and col_total_students:
    print("- Top 5 carreras por estudiantes (promedio):")
    show_top(df.groupby(col_carrera)[col_total_students].mean().sort_values(ascending=False))
if col_carrera and cost_col:
    print("- Top 5 carreras más costosas (promedio):")
    show_top(df.groupby(col_carrera)[cost_col].mean().sort_values(ascending=False))
    print("- Top 5 carreras más baratas (promedio):")
    show_top(df.groupby(col_carrera)[cost_col].mean().sort_values(ascending=True))

print("\n✅ EDA finalizado. Los CSV exportados quedan en el entorno de Colab.")
```
